```python
import os
import numpy as np
import pandas as pd

# Reuse the audio inventory
CSV_PATH = "/content/USEF-TSE-Dataset/metadata/audio_inventory.csv"
df = pd.read_csv(CSV_PATH)

# Only use recordings that are at least 4 seconds long
eligible_audio = df[df["duration"] >= 4.0].copy()

# Dataset sizes
mixture_counts = {
    "train-clean-100": 2000,
    "dev-clean": 300,
    "test-clean": 300
}

# Speaker counts per mixture
speaker_counts = [2, 3, 4]

# Fixed random seed for reproducibility
rng = np.random.default_rng(42)

plan = []

for split, count in mixture_counts.items():

    print("\nPreparing plan for:", split)

    subset = eligible_audio[eligible_audio["split"] == split]

    # Group recordings by speaker
    speaker_files = {
        speaker: group["audio_path"].tolist()
        for speaker, group in subset.groupby("speaker_id")
    }

    available_speakers = list(speaker_files.keys())

    for num_speakers in speaker_counts:

        print("Planning", count, "mixtures with",
              num_speakers, "speakers")

        for i in range(count):

            # Select different speakers for this mixture
            selected_speakers = rng.choice(
                available_speakers,
                size=num_speakers,
                replace=False
            )

            mixture_id = (
                f"{split}_{num_speakers}spk_{i:05d}"
            )

            for target_index, speaker in enumerate(selected_speakers):

                # Select two different recordings from the same speaker
                source_path, reference_path = rng.choice(
                    speaker_files[speaker],
                    size=2,
                    replace=False
                )

                plan.append({
                    "mixture_id": mixture_id,
                    "split": split,
                    "num_speakers": num_speakers,
                    "speaker_id": speaker,
                    "target_index": target_index,
                    "source_audio_path": source_path,
                    "reference_audio_path": reference_path
                })

# Convert to DataFrame
plan_df = pd.DataFrame(plan)

# Save the mixture plan
METADATA_DIR = "/content/USEF-TSE-Dataset/metadata"
os.makedirs(METADATA_DIR, exist_ok=True)

plan_path = os.path.join(
    METADATA_DIR,
    "mixture_plan.csv"
)

plan_df.to_csv(plan_path, index=False)

# Save a backup in Google Drive
drive_plan_path = os.path.join(
    DRIVE_BACKUP,
    "mixture_plan.csv"
)

plan_df.to_csv(drive_plan_path, index=False)

# Display summary
print("\n========== MIXTURE PLAN SUMMARY ==========")

summary = (
    plan_df.groupby(["split", "num_speakers"])["mixture_id"]
    .nunique()
    .unstack(fill_value=0)
)

print(summary)

print("\nTotal unique mixtures:",
      plan_df["mixture_id"].nunique())

print("Total target-conditioned examples:",
      len(plan_df))

# Verify reference and source are different recordings
same_recording = (
    plan_df["source_audio_path"] ==
    plan_df["reference_audio_path"]
).sum()

print("Source/reference recording conflicts:",
      same_recording)

# Verify each mixture has the correct number of speakers
speaker_check = plan_df.groupby("mixture_id")["speaker_id"].nunique()

invalid_mixtures = (
    plan_df.groupby("mixture_id")["num_speakers"].first()
    != speaker_check
).sum()

print("Invalid mixtures:", invalid_mixtures)

print("\nPlan saved successfully!")
print("Colab:", plan_path)
print("Google Drive:", drive_plan_path)
```

```python
import os
import time
import numpy as np
import pandas as pd
import soundfile as sf

DATASET_DIR = "/content/USEF-TSE-Dataset"
METADATA_DIR = os.path.join(DATASET_DIR, "metadata")
GENERATED_DIR = os.path.join(DATASET_DIR, "generated")

DRIVE_BACKUP = "/content/drive/MyDrive/USEF-TSE-Backup"

PLAN_PATH = os.path.join(METADATA_DIR, "mixture_plan.csv")

SAMPLE_RATE = 16000
SEGMENT_DURATION = 4
SEGMENT_SAMPLES = SAMPLE_RATE * SEGMENT_DURATION

RANDOM_SEED = 42

# Target speech level before mixing
TARGET_RMS = 10 ** (-25 / 20)

# Random gain variation for each speaker
MIN_GAIN_DB = -5
MAX_GAIN_DB = 5

# Maximum allowed mixture peak
MAX_PEAK = 0.95

plan_df = pd.read_csv(PLAN_PATH)

rng = np.random.default_rng(RANDOM_SEED)

for split in ["train-clean-100", "dev-clean", "test-clean"]:
    folder = {
        "train-clean-100": "train",
        "dev-clean": "validation",
        "test-clean": "test"
    }[split]

    os.makedirs(os.path.join(GENERATED_DIR, folder), exist_ok=True)

print("Configuration loaded successfully!")
print("Sample rate:", SAMPLE_RATE)
print("Segment duration:", SEGMENT_DURATION, "seconds")
print("Samples per segment:", SEGMENT_SAMPLES)
print("Mixture plan rows:", len(plan_df))
```

```python
def create_audio_segment(audio_path, rng):
    audio, sr = sf.read(audio_path, dtype="float32")

    if sr != SAMPLE_RATE:
        raise ValueError(
            f"Incorrect sample rate in {audio_path}: {sr}"
        )

    if audio.ndim != 1:
        raise ValueError(
            f"Audio is not mono: {audio_path}"
        )

    if len(audio) < SEGMENT_SAMPLES:
        raise ValueError(
            f"Audio is shorter than 4 seconds: {audio_path}"
        )

    max_start = len(audio) - SEGMENT_SAMPLES

    start = int(rng.integers(0, max_start + 1))

    segment = audio[start:start + SEGMENT_SAMPLES].copy()

    return segment, start


def adjust_source_volume(segment, rng):
    rms = np.sqrt(np.mean(segment ** 2))

    if rms < 1e-8:
        raise ValueError("Silent audio segment detected.")

    random_gain_db = rng.uniform(
        MIN_GAIN_DB,
        MAX_GAIN_DB
    )

    random_gain = 10 ** (random_gain_db / 20)

    gain = (TARGET_RMS / rms) * random_gain

    return segment * gain


def mix_speakers(source_segments):
    # Add all clean speaker segments
    mixture = np.sum(source_segments, axis=0)

    peak = np.max(np.abs(mixture))

    if peak < 1e-8:
        raise ValueError("Generated mixture is silent.")

    # Apply one common scale to mixture and source stems
    common_scale = min(1.0, MAX_PEAK / peak)

    source_segments = source_segments * common_scale
    mixture = mixture * common_scale

    return mixture.astype(np.float32), source_segments.astype(np.float32)


print("Audio-generation functions created successfully!")
```

```python
generated_records = []

split_folder = {
    "train-clean-100": "train",
    "dev-clean": "validation",
    "test-clean": "test"
}

start_time = time.time()

mixture_groups = plan_df.groupby(
    ["split", "mixture_id"],
    sort=False
)

total_mixtures = len(mixture_groups)

print("Starting audio generation...")
print("Total mixtures:", total_mixtures)

for mixture_number, ((split, mixture_id), group) in enumerate(
    mixture_groups,
    start=1
):

    group = group.sort_values("target_index")

    num_speakers = int(group["num_speakers"].iloc[0])

    if len(group) != num_speakers:
        raise ValueError(
            f"Incorrect number of rows for {mixture_id}"
        )

    if sorted(group["target_index"].tolist()) != list(
        range(num_speakers)
    ):
        raise ValueError(
            f"Incorrect target indices for {mixture_id}"
        )

    source_segments = []
    source_starts = []
    reference_starts = []

    for _, row in group.iterrows():

        # Get a 4-second clean source segment
        source_segment, source_start = create_audio_segment(
            row["source_audio_path"], rng
        )

        # Adjust the speaker's volume
        source_segment = adjust_source_volume(
            source_segment, rng
        )

        source_segments.append(source_segment)
        source_starts.append(source_start)

        # Select a separate 4-second reference segment
        reference_audio, reference_sr = sf.read(
            row["reference_audio_path"],
            dtype="float32"
        )

        if reference_sr != SAMPLE_RATE:
            raise ValueError("Reference sample rate mismatch.")

        if reference_audio.ndim != 1:
            raise ValueError("Reference audio is not mono.")

        if len(reference_audio) < SEGMENT_SAMPLES:
            raise ValueError("Reference audio is too short.")

        max_reference_start = (
            len(reference_audio) - SEGMENT_SAMPLES
        )

        reference_start = int(
            rng.integers(0, max_reference_start + 1)
        )

        reference_starts.append(reference_start)

    source_segments = np.stack(source_segments)

    # Generate the mixture and clean source stems
    mixture, source_segments = mix_speakers(
        source_segments
    )

    folder = split_folder[split]

    output_folder = os.path.join(
        GENERATED_DIR,
        folder
    )

    output_path = os.path.join(
        output_folder,
        mixture_id + ".npz"
    )

    # Save mixture and all source stems together
    np.savez_compressed(
        output_path,
        mixture=mixture,
        sources=source_segments
    )

    # Save one metadata row for each target speaker
    for position, (_, row) in enumerate(group.iterrows()):

        record = row.to_dict()

        record["mixture_path"] = output_path
        record["source_start_sample"] = source_starts[position]
        record["reference_start_sample"] = reference_starts[position]
        record["reference_duration_samples"] = SEGMENT_SAMPLES

        generated_records.append(record)

    if mixture_number % 100 == 0:
        elapsed = time.time() - start_time

        print(
            f"Generated {mixture_number}/{total_mixtures} "
            f"mixtures | Time: {elapsed / 60:.1f} minutes"
        )

# Save the generated metadata
generated_df = pd.DataFrame(generated_records)

metadata_path = os.path.join(
    METADATA_DIR,
    "generated_manifest.csv"
)

generated_df.to_csv(metadata_path, index=False)

# Back up the metadata to Google Drive
drive_metadata_path = os.path.join(
    DRIVE_BACKUP,
    "generated_manifest.csv"
)

generated_df.to_csv(drive_metadata_path, index=False)

total_time = time.time() - start_time

print("\n========== GENERATION COMPLETED ==========")
print("Unique mixtures:", generated_df["mixture_id"].nunique())
print("Target-conditioned examples:", len(generated_df))
print("Generation time:", round(total_time / 60, 2), "minutes")
print("Colab metadata:", metadata_path)
print("Drive metadata:", drive_metadata_path)
```

```python
import soundfile as sf
import numpy as np
import torch

from torch.utils.data import Dataset


class USEFTSEDataset(Dataset):

    def __init__(self, dataframe):
        self.dataframe = dataframe.reset_index(drop=True)

    def __len__(self):
        return len(self.dataframe)

    def __getitem__(self, index):

        row = self.dataframe.iloc[index]

        # Load the mixture and all clean source stems
        data = np.load(
            row["mixture_path"],
            allow_pickle=False
        )

        mixture = data["mixture"].astype(np.float32)

        sources = data["sources"]

        # Select the clean target using target_index
        target_index = int(row["target_index"])

        target = sources[target_index].astype(np.float32)

        # Load the separate reference recording
        reference, sr = sf.read(
            row["reference_audio_path"],
            dtype="float32"
        )

        if sr != 16000:
            raise ValueError("Reference sample rate is not 16000 Hz")

        if reference.ndim != 1:
            raise ValueError("Reference audio must be mono")

        # Select the previously recorded 4-second reference crop
        reference_start = int(row["reference_start_sample"])

        reference = reference[
            reference_start:reference_start + 64000
        ]

        if len(reference) != 64000:
            raise ValueError("Reference crop is not 4 seconds long")

        return {
            "mixture": torch.from_numpy(mixture.copy()),
            "reference": torch.from_numpy(reference.copy()),
            "target": torch.from_numpy(target.copy()),
            "num_speakers": int(row["num_speakers"])
        }


print("PyTorch Dataset class created successfully!")
```

```python
BATCH_SIZE = 4

train_dataset = USEFTSEDataset(train_df)
val_dataset = USEFTSEDataset(val_df)
test_dataset = USEFTSEDataset(test_df)

train_loader = DataLoader(
    train_dataset,
    batch_size=BATCH_SIZE,
    shuffle=True,
    num_workers=2,
    pin_memory=True,
    drop_last=True
)

val_loader = DataLoader(
    val_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

test_loader = DataLoader(
    test_dataset,
    batch_size=BATCH_SIZE,
    shuffle=False,
    num_workers=2,
    pin_memory=True
)

print("DataLoaders created successfully!")
print("Training batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Testing batches:", len(test_loader))
```

```python
import torch
import torch.nn as nn
import torch.nn.functional as F
import numpy as np
import time
import os

DEVICE = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

CONFIG = {
    "sample_rate": 16000,
    "encoder_channels": 64,
    "kernel_size": 16,
    "stride": 8,
    "attention_heads": 4,
    "cmha_layers": 2,
    "reference_tokens": 256,
    "chunk_size": 100,
    "intra_layers": 2,
    "inter_layers": 2,
    "dual_path_repeats": 2
}

print("========== MODEL CONFIGURATION ==========")

for key, value in CONFIG.items():
    print(key, ":", value)

print("\nDevice:", DEVICE)

if torch.cuda.is_available():
    print("GPU:", torch.cuda.get_device_name(0))
```

```python
class SharedEncoder(nn.Module):

    def __init__(self):
        super().__init__()

        self.encoder = nn.Conv1d(
            in_channels=1,
            out_channels=CONFIG["encoder_channels"],
            kernel_size=CONFIG["kernel_size"],
            stride=CONFIG["stride"]
        )

    def forward(self, audio):

        audio = audio.unsqueeze(1)

        features = self.encoder(audio)

        features = F.relu(features)

        return features


encoder = SharedEncoder().to(DEVICE)

print("Shared Encoder created!")
```

```python
class CrossAttentionBlock(nn.Module):

    def __init__(self, channels, heads):

        super().__init__()

        self.attention = nn.MultiheadAttention(
            embed_dim=channels,
            num_heads=heads,
            batch_first=True
        )

        self.norm1 = nn.LayerNorm(channels)

        self.feed_forward = nn.Sequential(
            nn.Linear(channels, channels * 2),
            nn.ReLU(),
            nn.Linear(channels * 2, channels)
        )

        self.norm2 = nn.LayerNorm(channels)

    def forward(self, query, key_value):

        attention_output, _ = self.attention(
            query,
            key_value,
            key_value,
            need_weights=False
        )

        output = self.norm1(
            query + attention_output
        )

        feed_forward_output = self.feed_forward(output)

        output = self.norm2(
            output + feed_forward_output
        )

        return output


class CMHA(nn.Module):

    def __init__(self):

        super().__init__()

        self.blocks = nn.ModuleList([
            CrossAttentionBlock(
                CONFIG["encoder_channels"],
                CONFIG["attention_heads"]
            )
            for _ in range(CONFIG["cmha_layers"])
        ])

    def forward(self, mixture_features, reference_features):

        # [B, C, L] -> [B, L, C]
        query = mixture_features.transpose(1, 2)

        # Reduce reference time steps to 256 tokens
        reference_features = F.adaptive_avg_pool1d(
            reference_features,
            CONFIG["reference_tokens"]
        )

        # [B, C, 256] -> [B, 256, C]
        key_value = reference_features.transpose(1, 2)

        for block in self.blocks:
            query = block(query, key_value)

        # [B, L, C] -> [B, C, L]
        return query.transpose(1, 2)


cmha = CMHA().to(DEVICE)

print("CMHA module created!")
```

```python
class FeatureFusion(nn.Module):

    def __init__(self, channels):

        super().__init__()

        self.gamma = nn.Linear(channels, channels)
        self.beta = nn.Linear(channels, channels)

    def forward(self, mixture_features, speaker_features):

        mixture_features = mixture_features.transpose(1, 2)
        speaker_features = speaker_features.transpose(1, 2)

        gamma = self.gamma(speaker_features)
        beta = self.beta(speaker_features)

        fused = gamma * mixture_features + beta

        return fused.transpose(1, 2)


fusion = FeatureFusion(
    CONFIG["encoder_channels"]
).to(DEVICE)

print("Feature Fusion module created!")
```

```python
class DualPathBlock(nn.Module):

    def __init__(self, channels):

        super().__init__()

        self.intra = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=channels,
                nhead=CONFIG["attention_heads"],
                dim_feedforward=channels * 2,
                batch_first=True,
                dropout=0.0
            ),
            num_layers=CONFIG["intra_layers"]
        )

        self.inter = nn.TransformerEncoder(
            nn.TransformerEncoderLayer(
                d_model=channels,
                nhead=CONFIG["attention_heads"],
                dim_feedforward=channels * 2,
                batch_first=True,
                dropout=0.0
            ),
            num_layers=CONFIG["inter_layers"]
        )

    def forward(self, x):

        batch, length, channels = x.shape

        chunk_size = CONFIG["chunk_size"]

        padding = (
            chunk_size - length % chunk_size
        ) % chunk_size

        if padding > 0:
            x = F.pad(x, (0, 0, 0, padding))

        padded_length = x.shape[1]

        num_chunks = padded_length // chunk_size

        # Split into chunks
        x = x.reshape(
            batch,
            num_chunks,
            chunk_size,
            channels
        )

        # Intra-chunk processing
        intra_input = x.reshape(
            batch * num_chunks,
            chunk_size,
            channels
        )

        intra_output = self.intra(intra_input)

        x = intra_output.reshape(
            batch,
            num_chunks,
            chunk_size,
            channels
        )

        # Inter-chunk processing
        inter_input = x.permute(
            0, 2, 1, 3
        ).reshape(
            batch * chunk_size,
            num_chunks,
            channels
        )

        inter_output = self.inter(inter_input)

        x = inter_output.reshape(
            batch,
            chunk_size,
            num_chunks,
            channels
        ).permute(
            0, 2, 1, 3
        )

        # Reconstruct the original sequence
        x = x.reshape(
            batch,
            padded_length,
            channels
        )

        return x[:, :length, :]


class Separator(nn.Module):

    def __init__(self, channels):

        super().__init__()

        self.blocks = nn.ModuleList([
            DualPathBlock(channels)
            for _ in range(CONFIG["dual_path_repeats"])
        ])

        self.mask_layer = nn.Conv1d(
            channels,
            channels,
            kernel_size=1
        )

    def forward(self, fused_features, mixture_features):

        x = fused_features.transpose(1, 2)

        for block in self.blocks:
            x = block(x)

        x = x.transpose(1, 2)

        mask = torch.sigmoid(
            self.mask_layer(x)
        )

        separated = mixture_features * mask

        return separated


separator = Separator(
    CONFIG["encoder_channels"]
).to(DEVICE)

print("SepFormer-style Separator created!")
```

```python
class Decoder(nn.Module):

    def __init__(self):

        super().__init__()

        self.decoder = nn.ConvTranspose1d(
            in_channels=CONFIG["encoder_channels"],
            out_channels=1,
            kernel_size=CONFIG["kernel_size"],
            stride=CONFIG["stride"]
        )

    def forward(self, features):

        audio = self.decoder(features)

        return audio.squeeze(1)


decoder = Decoder().to(DEVICE)

print("Decoder created!")
```

```python
class USEFTSE(nn.Module):

    def __init__(self):

        super().__init__()

        self.encoder = SharedEncoder()
        self.cmha = CMHA()
        self.fusion = FeatureFusion(
            CONFIG["encoder_channels"]
        )
        self.separator = Separator(
            CONFIG["encoder_channels"]
        )
        self.decoder = Decoder()

    def forward(self, mixture, reference):

        # Encode both audio inputs using shared weights
        mixture_features = self.encoder(mixture)
        reference_features = self.encoder(reference)

        # Identify target-speaker features
        speaker_features = self.cmha(
            mixture_features,
            reference_features
        )

        # Combine mixture and speaker features
        fused_features = self.fusion(
            mixture_features,
            speaker_features
        )

        # Separate target-speaker features
        separated_features = self.separator(
            fused_features,
            mixture_features
        )

        # Reconstruct target audio
        estimated_audio = self.decoder(
            separated_features
        )

        return estimated_audio


model = USEFTSE().to(DEVICE)

total_parameters = sum(
    parameter.numel()
    for parameter in model.parameters()
)

print("USEF-TSE model created successfully!")
print("Total parameters:", f"{total_parameters:,}")
print("Model device:", DEVICE)
```

```python
def si_sdr_loss(estimate, target, eps=1e-8):

    # Remove the mean from both signals
    estimate = estimate - estimate.mean(dim=-1, keepdim=True)
    target = target - target.mean(dim=-1, keepdim=True)

    # Project estimate onto the clean target
    target_energy = torch.sum(
        target ** 2, dim=-1, keepdim=True
    ) + eps

    scale = torch.sum(
        estimate * target, dim=-1, keepdim=True
    ) / target_energy

    projected = scale * target

    # Calculate the noise/residual
    noise = estimate - projected

    target_power = torch.sum(
        projected ** 2, dim=-1
    ) + eps

    noise_power = torch.sum(
        noise ** 2, dim=-1
    ) + eps

    si_sdr = 10 * torch.log10(
        target_power / noise_power
    )

    return -si_sdr.mean()


print("SI-SDR loss function created!")
```

```python
import os
import time
import pandas as pd
import torch
import numpy as np

EPOCHS = 10
LEARNING_RATE = 0.0001
PATIENCE = 3

BATCH_SIZE = 4

CHECKPOINT_DIR = (
    "/content/drive/MyDrive/USEF-TSE-Backup/checkpoints"
)

LOG_DIR = (
    "/content/drive/MyDrive/USEF-TSE-Backup/training_logs"
)

os.makedirs(CHECKPOINT_DIR, exist_ok=True)
os.makedirs(LOG_DIR, exist_ok=True)

optimizer = torch.optim.Adam(
    model.parameters(),
    lr=LEARNING_RATE
)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode="min",
    factor=0.5,
    patience=2
)

history = []

best_val_loss = float("inf")
epochs_without_improvement = 0

print("========== TRAINING CONFIGURATION ==========")
print("Epochs:", EPOCHS)
print("Batch size:", BATCH_SIZE)
print("Learning rate:", LEARNING_RATE)
print("Optimizer: Adam")
print("Loss: Negative SI-SDR")
print("Early stopping patience:", PATIENCE)
print("Checkpoint directory:", CHECKPOINT_DIR)
```

```python
def run_training_epoch(model, loader, optimizer):

    model.train()

    total_loss = 0.0
    total_examples = 0

    for batch_index, batch in enumerate(loader, start=1):

        mix = batch["mixture"].to(DEVICE, non_blocking=True)
        ref = batch["reference"].to(DEVICE, non_blocking=True)
        target = batch["target"].to(DEVICE, non_blocking=True)

        optimizer.zero_grad(set_to_none=True)

        estimate = model(mix, ref)

        loss = si_sdr_loss(estimate, target)

        if not torch.isfinite(loss):
            raise RuntimeError(
                f"Non-finite training loss at batch {batch_index}"
            )

        loss.backward()

        torch.nn.utils.clip_grad_norm_(
            model.parameters(),
            max_norm=5.0
        )

        optimizer.step()

        batch_size = mix.size(0)

        total_loss += loss.item() * batch_size
        total_examples += batch_size

        if batch_index % 500 == 0:
            print(
                f"Training batch {batch_index}/{len(loader)}"
            )

    return total_loss / total_examples


def run_validation_epoch(model, loader):

    model.eval()

    total_loss = 0.0
    total_examples = 0

    with torch.no_grad():

        for batch in loader:

            mix = batch["mixture"].to(DEVICE, non_blocking=True)
            ref = batch["reference"].to(DEVICE, non_blocking=True)
            target = batch["target"].to(DEVICE, non_blocking=True)

            estimate = model(mix, ref)

            loss = si_sdr_loss(estimate, target)

            if not torch.isfinite(loss):
                raise RuntimeError("Non-finite validation loss")

            batch_size = mix.size(0)

            total_loss += loss.item() * batch_size
            total_examples += batch_size

    return total_loss / total_examples


overall_start = time.time()

for epoch in range(1, EPOCHS + 1):

    epoch_start = time.time()

    if DEVICE.type == "cuda":
        torch.cuda.reset_peak_memory_stats()

    print(f"\n========== EPOCH {epoch}/{EPOCHS} ==========")

    train_loss = run_training_epoch(
        model,
        train_loader,
        optimizer
    )

    val_loss = run_validation_epoch(
        model,
        val_loader
    )

    scheduler.step(val_loss)

    epoch_time = time.time() - epoch_start

    if DEVICE.type == "cuda":
        peak_memory = (
            torch.cuda.max_memory_allocated() / (1024 ** 3)
        )
    else:
        peak_memory = 0.0

    current_lr = optimizer.param_groups[0]["lr"]

    history.append({
        "epoch": epoch,
        "train_loss": train_loss,
        "validation_loss": val_loss,
        "epoch_time_seconds": epoch_time,
        "gpu_peak_memory_gb": peak_memory,
        "learning_rate": current_lr
    })

    history_df = pd.DataFrame(history)

    history_df.to_csv(
        os.path.join(LOG_DIR, "training_history.csv"),
        index=False
    )

    print("\nEpoch results:")
    print("Training loss:", round(train_loss, 4))
    print("Validation loss:", round(val_loss, 4))
    print("Epoch time:", round(epoch_time / 60, 2), "minutes")
    print("Peak GPU memory:", round(peak_memory, 3), "GB")
    print("Learning rate:", current_lr)

    # Save the best model
    if val_loss < best_val_loss:

        best_val_loss = val_loss
        epochs_without_improvement = 0

        checkpoint = {
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "best_val_loss": best_val_loss,
            "config": CONFIG,
            "history": history
        }

        torch.save(
            checkpoint,
            os.path.join(CHECKPOINT_DIR, "best_model.pth")
        )

        print("Best model checkpoint saved!")

    else:
        epochs_without_improvement += 1
        print(
            "No validation improvement:",
            epochs_without_improvement,
            "/", PATIENCE
        )

    # Save latest model separately
    torch.save(
        {
            "epoch": epoch,
            "model_state_dict": model.state_dict(),
            "optimizer_state_dict": optimizer.state_dict(),
            "config": CONFIG,
            "history": history
        },
        os.path.join(CHECKPOINT_DIR, "latest_model.pth")
    )

    if epochs_without_improvement >= PATIENCE:
        print("\nEarly stopping triggered.")
        break


total_training_time = time.time() - overall_start

print("\n========== TRAINING COMPLETED ==========")
print("Epochs completed:", len(history))
print("Best validation loss:", best_val_loss)
print(
    "Total training time:",
    round(total_training_time / 60, 2),
    "minutes"
)
print("Training history saved:", LOG_DIR)
print("Model checkpoints saved:", CHECKPOINT_DIR)
```

```python
def calculate_si_sdr(estimate, target, eps=1e-8):

    estimate = estimate - estimate.mean(dim=-1, keepdim=True)
    target = target - target.mean(dim=-1, keepdim=True)

    target_energy = torch.sum(target ** 2, dim=-1, keepdim=True) + eps

    scale = torch.sum(
        estimate * target, dim=-1, keepdim=True
    ) / target_energy

    projected = scale * target
    noise = estimate - projected

    target_power = torch.sum(projected ** 2, dim=-1) + eps
    noise_power = torch.sum(noise ** 2, dim=-1) + eps

    return 10 * torch.log10(target_power / noise_power)


print("SI-SDR evaluation function created!")
```

```python
import numpy as np
import pandas as pd
import torch
import os

model.eval()

results = {
    2: {"si_sdr": [], "mixture_si_sdr": [], "si_sdri": []},
    3: {"si_sdr": [], "mixture_si_sdr": [], "si_sdri": []},
    4: {"si_sdr": [], "mixture_si_sdr": [], "si_sdri": []}
}

total_tested = 0

with torch.no_grad():

    for batch_number, batch in enumerate(test_loader, start=1):

        mix = batch["mixture"].to(DEVICE)
        ref = batch["reference"].to(DEVICE)
        target = batch["target"].to(DEVICE)

        num_speakers = batch["num_speakers"]

        estimated = model(mix, ref)

        estimated_si_sdr = calculate_si_sdr(estimated, target)
        mixture_si_sdr = calculate_si_sdr(mix, target)

        si_sdri = estimated_si_sdr - mixture_si_sdr

        speaker_counts = num_speakers.tolist()

        for i, count in enumerate(speaker_counts):

            count = int(count)

            results[count]["si_sdr"].append(
                estimated_si_sdr[i].item()
            )

            results[count]["mixture_si_sdr"].append(
                mixture_si_sdr[i].item()
            )

            results[count]["si_sdri"].append(
                si_sdri[i].item()
            )

        total_tested += mix.size(0)

        if batch_number % 100 == 0:
            print(
                f"Test batches completed: "
                f"{batch_number}/{len(test_loader)}"
            )

print("\nTesting completed!")
print("Total test examples:", total_tested)
```

```python
import torch
import numpy as np
import pandas as pd

model.eval()

SAMPLE_RATE = 16000

# Reference durations to test
reference_durations = {
    "4 seconds": 4 * SAMPLE_RATE,
    "2 seconds": 2 * SAMPLE_RATE,
    "1 second": 1 * SAMPLE_RATE,
    "0.5 seconds": int(0.5 * SAMPLE_RATE)
}

short_reference_results = {}

for duration_name, num_samples in reference_durations.items():

    print(f"\nTesting reference duration: {duration_name}")

    results = {
        2: {"si_sdr": [], "si_sdri": []},
        3: {"si_sdr": [], "si_sdri": []},
        4: {"si_sdr": [], "si_sdri": []}
    }

    with torch.no_grad():

        for batch in test_loader:

            mix = batch["mixture"].to(DEVICE)
            ref = batch["reference"].to(DEVICE)
            target = batch["target"].to(DEVICE)

            speaker_counts = batch["num_speakers"].tolist()

            # Keep only the required reference duration
            short_ref = ref[:, :num_samples]

            # Extract target speaker
            estimated = model(mix, short_ref)

            # Calculate SI-SDR
            estimated_si_sdr = calculate_si_sdr(
                estimated, target
            )

            mixture_si_sdr = calculate_si_sdr(
                mix, target
            )

            si_sdri = estimated_si_sdr - mixture_si_sdr

            for i, count in enumerate(speaker_counts):

                count = int(count)

                results[count]["si_sdr"].append(
                    estimated_si_sdr[i].item()
                )

                results[count]["si_sdri"].append(
                    si_sdri[i].item()
                )

    short_reference_results[duration_name] = results

    print(f"Completed testing: {duration_name}")

print("\nAll reference duration experiments completed!")
```
